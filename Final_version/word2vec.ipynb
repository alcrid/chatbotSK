```python
import methods as md
import multiprocessing
from gensim.models import Word2Vec
cores = multiprocessing.cpu_count() #
import time
import numpy as np
```

```python
s = "D:\\Datasets\\facebook\\facebook-json-all-low"
dataset = md.make_datasetv2(s)
text = []


for x in dataset:
    sent = []
    for word in (x[0]+" " + x[1]).split(" "):
        if(word != ""):
            sent.append(word)
    text.append(sent) 

```

```python
word_freq = {}
for sent in text:
    for i in sent:
        if(i in word_freq):
            word_freq[i] += 1
        else:
            word_freq[i] = 1

word_freq["<NOT>"] = 5
word_freq["<PAD>"] = 5
word_freq["<SOS>"] = 5
word_freq["<EOS>"] = 5
```

```python
w2v_model = Word2Vec(min_count=1,
                     window=3,
                     vector_size=256,
                     sample=6e-5, 
                     alpha=0.03, 
                     min_alpha=0.0007, 
                     negative=20,
                     workers=cores-1)
```

```python

t = time.time()

w2v_model.build_vocab(text, progress_per=10000)

print('Time to build vocab: {} mins'.format(round((time.time() - t) / 60, 2)))
```

```python
t = time.time()
w2v_model.train(list(word_freq), total_examples=w2v_model.corpus_count, epochs=30, report_delay=1)

```

```python
w2v_model.init_sims(replace=True)
```

```python
w2v_model.wv.most_similar(positive=["teraz"])

```

```python
w2v_model.save("word2vec2.model")
```
